# Percorrer e filtrar

Quais temperaturas devem continuar se o filtro aceita apenas valores acima de zero? Retome [For e acumuladores](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/for-e-acumuladores) e [Listas e cópias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/listas).

`for valor in dados:` entrega os elementos, um por vez. Se você precisa também da posição, `for indice, valor in enumerate(dados):` dá os dois valores de cada passagem, com índice começando em 0. A condição dentro do laço decide quais itens entram em uma lista inicialmente vazia. Use append na lista de resultado para manter a entrada intacta.

Neste conjunto, `None` representa um valor ausente. `valor is None` testa a presença desse marcador. Um teste como `not valor` englobaria também zero e confundiria dois estados diferentes.

Em [Compreensões](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/compreensoes) você vai reconhecer esse mesmo filtro em uma expressão. Antes, explique o que se perderia se o teste tratasse zero como ausência.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-percorrer-e-filtrar-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja quais itens de [−1, 2, 0, 4] passam pelo limite estrito. Construa uma lista contendo só temperaturas positivas.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
temperaturas = [-1, 2, 0, 4]
positivas = []
for temperatura in temperaturas:
    if temperatura > 0:
        positivas.append(temperatura)
print(positivas)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

```

A lista começa vazia. O teste > 0 exclui −1 e 0. Cada valor aceito entra na ordem em que foi encontrado.

## E3 · Resolva

Localize os índices de valores ausentes em [3, None, 0, None]. Use None como único marcador de ausência e mantenha a lista original. Mostre os índices. Depois teste [0], que deve produzir lista de índices vazia.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
medicoes = [3, None, 0, None]
ausentes = []
for indice, valor in enumerate(medicoes):
    if valor is None:
        ausentes.append(indice)
print(ausentes)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
[1, 3]
```

Guardamos o índice quando o item é None. Zero é uma medição presente. A lista [0] tem uma medição e nenhum índice de ausência.
