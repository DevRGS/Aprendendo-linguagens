# Listas e cópias

Se você acrescentar 8 a uma lista de medidas, o que verá pelo nome original? Retome [Texto e fatias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/texto-e-fatias) para rever o índice inicial 0.

Uma lista ordena itens entre colchetes. Em `medidas = [4, 6]`, `medidas[0]` é 4. `medidas.append(8)` acrescenta ao fim e altera a lista existente. `medidas[1] = 7` substitui o segundo item. `medidas.remove(7)` remove a primeira ocorrência do valor 7. Para remover pela posição, use `pop(indice)`, que também devolve o item removido.

`outra = medidas` liga outro nome à mesma lista. Uma alteração feita por qualquer nome aparece pelo outro. `outra = medidas.copy()` cria uma nova lista externa, de modo que substituir ou acrescentar itens nela não altera a lista original. Essa cópia é rasa: se os itens forem listas internas, elas continuam compartilhadas. Aqui usamos listas planas de inteiros.

Antes de seguir para [Percorrer e filtrar](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/percorrer-e-filtrar), explique por que dar um novo nome à mesma lista não protege as medidas originais.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-listas-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja a posição de 8 depois de acrescentá-lo a [4, 6]. Mostre a lista resultante.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````
