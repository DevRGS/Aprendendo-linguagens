# Compreensões de lista

Como escrever o dobro de cada item em uma lista sem perder de vista o laço equivalente? Retome [Percorrer e filtrar](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/percorrer-e-filtrar).

O laço começa com uma lista vazia, percorre os valores e acrescenta `valor * 2`. A compreensão `[valor * 2 for valor in dados]` produz uma nova lista com a mesma sequência. Leia primeiro o `for`, identifique cada valor e então aplique a expressão à esquerda. Em `[valor * 2 for valor in dados if valor % 2 == 0]`, o teste atua no valor original. Somente quando passa é que o dobro vai para a saída.

Para filtrar depois de transformar, guarde os valores transformados e use outra etapa. Manter duas etapas ajuda a enxergar qual valor passa pelo teste. Ao terminar, explique por que dobrar primeiro pode mudar o resultado. Na próxima aula você vai agrupar valores que pertencem ao mesmo registro.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-compreensoes-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o resultado de dobrar [1, 2, 3]. Execute o laço e a compreensão e compare as listas.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
dados = [1, 2, 3]
pelo_laco = []
for valor in dados:
    pelo_laco.append(valor * 2)

pela_compreensao = [valor * 2 for valor in dados]
print(pelo_laco)
print(pela_compreensao)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
[2, 4, 6]
[2, 4, 6]
```

Cada valor é lido uma vez em cada versão. A expressão da compreensão calcula o mesmo valor enviado a append.
