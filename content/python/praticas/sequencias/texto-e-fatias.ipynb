# Texto e fatias

Como transformar `" Ana "` no nome `ana` para compará-lo com outro cadastro? Retome [Nomes e tipos](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/nomes-e-tipos) para rever texto e atribuição.

O índice começa em zero: em `"sol"`, o índice 0 aponta para `s`. Uma fatia como `"sol"[0:2]` começa em 0 e vai até antes de 2, produzindo `"so"`. Uma fatia pode ser vazia. Um índice individual precisa existir. Texto é imutável. Você pode ligar um nome a um novo texto, mas não pode trocar um caractere do texto original por atribuição a um índice.

Métodos são operações acessadas depois de um ponto. Em `" Ana ".strip().lower()`, `strip()` produz `"Ana"` ao tirar espaços nas extremidades. `lower()` usa esse resultado e produz `"ana"`. O método `split(",")` separa campos onde encontra vírgulas e devolve uma lista. Na próxima aula você vai alterar os itens de uma lista. Antes, explique por que cortar uma fatia é diferente de editar o texto no lugar.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-texto-e-fatias-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o texto depois de cada método. Normalize o nome com espaços e maiúscula inicial.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````
