# Resumir dados por grupo

Duas estações têm o mesmo tempo médio. Uma delas tem uma viagem sem tempo registrado. Que contagens devem acompanhar as médias?

Use pandas. Os exercícios criam tabelas no código. Se necessário, retome [Resumo de lote](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/resumo-de-lote).

`groupby` separa linhas pela chave escolhida. Para uma coluna de medidas, `size` conta todas as linhas do grupo e `count` conta valores presentes. A média usa os valores presentes. Com `agg`, nomeie as medidas que serão apresentadas juntas. Ordenar pela chave facilita comparar as saídas.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-agregar-', dir=_raiz))
import pandas
````

## E1 · Preveja e execute

Agrupe A:2, A:4 e B:3. Calcule as médias à mão e depois confira.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import pandas as pd

df = pd.DataFrame({"estacao": ["A", "A", "B"], "minutos": [2, 4, 3]})
print(df.groupby("estacao")["minutos"].mean().to_dict())
````

## E3 · Resolva

Para B, A, B e C com tempos 3, 0, 5 e ausente, mostre linhas, presentes e média, em ordem de estação. Verifique C, sem tempos presentes.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import pandas as pd

df = pd.DataFrame({"estacao": ["B", "A", "B", "C"], "minutos": [3, 0, 5, None]})
tabela = df.groupby("estacao", sort=True)["minutos"].agg(
    linhas="size", presentes="count", media="mean"
)
for chave, linha in tabela.iterrows():
    media = "ausente" if pd.isna(linha["media"]) else f"{linha['media']:.1f}"
    print(chave, int(linha["linhas"]), int(linha["presentes"]), media)
assert pd.isna(tabela.loc["C", "media"])
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
A 1 1 0.0
B 2 2 4.0
C 1 0 ausente
```

O zero em A é uma observação. A média de C não é definida, pois o denominador é zero. Em seguida, use a chave de estação para adicionar um nome legível.
