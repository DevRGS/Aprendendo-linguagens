# Chaves e junções de tabelas

Anexe nomes de estação a uma tabela em que uma linha é uma viagem. O que deverá acontecer com uma viagem que usa uma estação sem cadastro?

Use pandas. Retome [Resumir dados por grupo](https://devrgs.github.io/Aprendendo-linguagens/#/python/pandas/agregar). As tabelas são criadas no código.

Uma chave relaciona registros. Muitas viagens podem apontar para a mesma estação, mas o cadastro deve ter no máximo uma linha por chave. `merge` com `how="left"` mantém as viagens do lado esquerdo. Um `inner` conserva somente linhas com par. `validate="many_to_one"` verifica a unicidade na direita. `indicator=True` cria a coluna `_merge`, que contém `left_only` para viagem sem par.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-juntar-tabelas-', dir=_raiz))
import pandas
````

## E1 · Preveja e execute

Combine duas viagens com o cadastro e confira que os IDs de viagem continuam na saída.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import pandas as pd

viagens = pd.DataFrame({"id": ["v1", "v2"], "estacao": ["A", "A"]})
estacoes = pd.DataFrame({"estacao": ["A", "B"], "nome": ["Centro", "Norte"]})
uniao = viagens.merge(estacoes, on="estacao", how="left", validate="many_to_one")
print(uniao[["id", "nome"]].values.tolist())
````
