# Ler e inspecionar tabelas

Dois IDs vêm escritos como 001 e 002. Preveja o que um leitor fará se interpretar a coluna como números.

Revise [CSV](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/csv). Use pandas com o kernel do curso ou com as dependências do kit. Cada exercício recebe o arquivo `dados/viagens.csv`. O script resolve o caminho a partir da sua pasta. No caderno a preparação seleciona essa pasta como diretório corrente.

Um `DataFrame` é uma tabela com índices e colunas nomeados. Uma coluna selecionada isoladamente é uma `Series`, com valores, índice e tipo próprios. `pd.read_csv` usa o cabeçalho para nomear as colunas e tenta inferir tipos quando você não os declara. `shape` informa (linhas, colunas), `dtypes` descreve cada coluna e `head(n)` oferece uma amostra das primeiras n linhas.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-ler-e-inspecionar-', dir=_raiz))
import pandas
````

## E1 · Preveja e execute

Leia o arquivo com o ID como texto para preservar os zeros. Mostre os IDs e a primeira linha.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'dados/viagens.csv'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('id,km,minutos\n001,2.5,7\n002,7.0,19\n', encoding='utf-8')
_refresh_imports()
````
