# Composição de funções

Onde investigar se um relatório mostra o nome certo mas a soma errada? Você vai passar o resultado de uma função à seguinte e testar cada etapa. Retome [Contratos e tipos](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/contratos-e-tipos) e [Resumo de lote](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/resumo-de-lote).

Separe a obtenção dos dados, a transformação e a apresentação. Nesta aula os dados de entrada já estão em uma lista. No começo da composição, `limpar_nome(" Lia ")` devolve `"Lia"`. Em seguida `gerar_etiqueta("Lia")` devolve `"Nome: Lia"`. O código que chama decide quando imprimir. A saída de uma etapa precisa respeitar o formato que a seguinte recebe.

Funções também podem ser argumentos. Em `sorted(registros, key=duracao)`, passe a função pelo nome, sem parênteses. sorted chama `duracao` para cada registro e usa seu retorno para ordenar. Chamar `duracao(registro)` no lugar do argumento key passaria um valor e perderia a regra.

Teste limpeza sem somar e teste soma com nomes já limpos. Se cada etapa tiver uma entrada conhecida, você consegue descobrir em qual delas o resultado mudou. Na próxima aula a entrada vai chegar de um arquivo.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-composicao-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o retorno da limpeza e a etiqueta final para um nome com espaços. Execute a composição e exiba a etiqueta.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def limpar_nome(texto):
    return texto.strip()

def gerar_etiqueta(nome):
    return f"Nome: {nome}"

print(gerar_etiqueta(limpar_nome(" Lia ")))
````
