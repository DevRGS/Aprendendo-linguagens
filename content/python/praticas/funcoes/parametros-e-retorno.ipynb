# Parâmetros e retorno

Como aproveitar o dobro de 3 em outra conta? Você vai criar uma função para calcular o valor e deixar a chamada decidir como usá-lo. Retome [Contas e formatação](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/contas-e-formatacao) para rever expressões e [Condições e faixas](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/condicoes) para rever indentação.

`def dobro(numero):` define a função e declara um parâmetro chamado `numero`. Ao chamar `dobro(3)`, o 3 é o argumento e `numero` é o nome que o recebe dentro do bloco. `return numero * 2` termina a chamada e entrega 6 ao chamador. Assim `dobro(3) + 1` pode calcular 7. Rode a definição antes da chamada, no mesmo arquivo ou célula.

`print` escreve na saída. Uma função que chega ao fim do bloco sem executar `return` devolve `None`. Compare o que aparece na tela com o valor que a chamada entrega. Antes de passar para [Escopo e padrões](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/escopo-e-padroes), explique onde colocaria a impressão se quisesse reutilizar a função num cálculo silencioso.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-parametros-e-retorno-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o retorno de dobro(3) e o resultado ao acrescentar 1. Mostre os dois.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def dobro(numero):
    return numero * 2

resultado = dobro(3)
print(resultado)
print(resultado + 1)
````
