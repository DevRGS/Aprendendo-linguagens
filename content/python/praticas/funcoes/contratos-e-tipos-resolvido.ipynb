# Contratos e tipos

Como comunicar a quem chama uma função de média o que fazer com uma lista vazia? Retome [Parâmetros e retorno](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno) e [Listas e cópias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/listas).

Um contrato explicita pré-condições, como “a lista contém números e tem pelo menos um valor”, e pós-condições, como “o resultado é a soma dividida pela quantidade”. Uma docstring é o texto colocado como primeira instrução do corpo. Use-a para descrever a regra e limites que a assinatura não expressa sozinha.

Em `def media(valores: list[float]) -> float:`, `:` depois de `valores` anota o tipo esperado do argumento e `->` anota o retorno. Anotar não insere testes de validade em tempo de execução. Uma lista vazia chega ao corpo, e dividir por seu comprimento zero levanta `ZeroDivisionError`. Uma função pode escolher outro contrato, como retornar `None` no caso vazio. A anotação `float | None` permite comunicar essas duas possibilidades.

Antes de passar para [Composição](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/composicao), compare a média de [0, 0] com o resultado de []. Uma tem valor numérico e a outra representa ausência.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-contratos-e-tipos-', dir=_raiz))
````

## E1 · Preveja e execute

A função abaixo espera uma lista numérica não vazia. Preveja a soma, a quantidade e o retorno para [2.0, 4.0].


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def media(valores: list[float]) -> float:
    """Calcule a média de uma lista não vazia."""
    return sum(valores) / len(valores)

print(media([2.0, 4.0]))
````

sum calcula 6.0 e len informa 2. A divisão retorna 3.0, conforme a regra documentada.

## E3 · Resolva

Implemente `media_ou_none` com anotação e docstring. Para uma lista de números não vazia, devolva a média como float. Para uma vazia, devolva None. Mostre o resultado de [2, 4] e []. Confira o caso [0, 0] para distinguir zero de ausência.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def media_ou_none(valores: list[float]) -> float | None:
    """Retorne a média dos números ou None quando a lista estiver vazia."""
    if len(valores) == 0:
        return None
    return sum(valores) / len(valores)

print(media_ou_none([2, 4]))
print(media_ou_none([]))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
3.0
None
```

A função verifica o limite antes da divisão. Uma lista de dois zeros produz 0.0 e não passa pelo ramo da ausência. As anotações descrevem os tipos previstos e o teste de comprimento faz a decisão real.
