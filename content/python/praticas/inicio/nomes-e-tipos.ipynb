# Nomes e tipos

Como guardar 4 aulas e acrescentar 2 antes de exibir o total? Você já precisa conseguir chamar `print`. Se quiser retomar, abra [Primeira execução](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/primeira-execucao). Cada prática começa com os próprios valores.

Um literal é um valor escrito diretamente no código. Em `aulas = 4`, o literal inteiro é 4 e a atribuição liga o nome `aulas` a esse valor. Ao ler `aulas = aulas + 2`, comece pelo lado direito: o valor anterior é 4, a soma produz 6 e só depois o nome passa a designar 6.

`int` é o tipo de 4, `float` é o tipo de 4.0, `str` é o tipo de `"4"` e `bool` é o tipo dos valores `True` e `False`. O tipo define operações possíveis. Multiplicar um texto por um inteiro repete o texto. Multiplicar dois inteiros faz uma conta. Em `type(valor).__name__`, `type(valor)` identifica o tipo e `.__name__` extrai seu nome curto.

Preveja se `print("4")` e `print(4)` exibem a mesma linha. A saída pode coincidir mesmo quando os tipos diferem. A próxima aula usa números para converter unidades.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-nomes-e-tipos-', dir=_raiz))
````

## E1 · Preveja e execute

Antes de executar, diga quanto vale `aulas` depois de cada atribuição. Exiba o total depois de acrescentar duas aulas a quatro.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
aulas = 4
aulas = aulas + 2
print(aulas)
````


A primeira atribuição liga aulas a 4. A expressão à direita da segunda usa esse 4 e liga aulas ao resultado 6.
