# Contas e formatação

Quantas horas completas cabem em 135 minutos e quanto sobra? Você vai escolher operações e exibir o resultado em uma forma legível. Retome [Nomes e tipos](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/nomes-e-tipos) se precisar rever atribuições e números.

Use `+`, `-` e `*` para somar, subtrair e multiplicar. Multiplicação e divisão têm prioridade sobre soma e subtração. Parênteses indicam a operação que você quer calcular primeiro. Para inteiros como 135 e 60, `135 / 60` devolve 2.25, um `float`. `135 // 60` dá o quociente inteiro 2 e `135 % 60` dá o resto 15. Conferimos com `2 * 60 + 15 == 135`.

Uma f-string começa com `f` e deixa inserir valores calculados entre chaves. Se `valor` é 7.5, `f"{valor:.2f}"` produz o texto `7.50`. A instrução após os dois pontos controla a apresentação. O número guardado em `valor` continua 7.5.

Antes de avançar, explique qual operação mostraria apenas os minutos que sobraram. Na próxima aula, o valor virá de uma entrada digitada.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-contas-e-formatacao-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o número de horas completas e de minutos restantes. Em seguida, converta 135 minutos e imprima os dois valores em uma frase.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````
