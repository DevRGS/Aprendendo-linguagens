# For e acumuladores

Qual será o total depois de cada adição de 1 até 4? Você vai percorrer números e guardar o resultado parcial. Retome [Condições e faixas](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/condicoes) para rever blocos indentados.

O `for` entrega um valor por iteração, ou repetição. Em `for numero in range(1, 5):`, o primeiro valor é 1 e o limite final 5 fica de fora. O corpo é o bloco indentado. Comece o acumulador `total` em 0 antes do laço. Na passagem com número 2, o total antes da soma é 1 e depois vira 3. Imprimir dentro do laço mostra a evolução. Imprimir depois do laço mostra só o resultado final.

Com três argumentos, `range(inicio, fim, passo)` avança pelo passo indicado. Preveja o primeiro e o último valor de `range(2, 11, 2)`. Em seguida você vai controlar a repetição com uma condição de parada.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-for-e-acumuladores-', dir=_raiz))
````

## E1 · Preveja e execute

Anote os quatro totais esperados antes de executar. Some 1 até 4 e mostre o acumulador depois de cada adição.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
total = 0
for numero in range(1, 5):
    total = total + numero
    print(total)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada


A cada iteração, o valor anterior de total participa da nova soma. Colocar total = 0 dentro do laço apagaria os resultados parciais.
