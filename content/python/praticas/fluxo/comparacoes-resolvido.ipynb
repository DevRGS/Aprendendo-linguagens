# Comparações e lógica

A nota zero deve passar por uma validação que aceita valores de 0 a 10? Use nomes e valores como em [Nomes e tipos](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/nomes-e-tipos) para escrever regras que devolvem `True` ou `False`.

Em `nota >= 0`, o sinal `>=` inclui a igualdade. Use `==` para comparar valores e `=` para atribuir um valor a um nome. `0 <= nota <= 10` verifica os dois lados do intervalo: 0 e 10 pertencem ao intervalo e 11 fica fora. Para condições booleanas, `and` exige as duas verdadeiras, `or` aceita uma ou ambas e `not` inverte a condição. Em `True or False and False`, `and` é avaliado como parte do operando à direita de `or`. A expressão é verdadeira. Parênteses permitem exigir outra associação.

Explique por que um limite de “pelo menos 16” inclui 16. Em [Condições](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/condicoes), as comparações vão decidir quais linhas executar.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-comparacoes-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja se as notas 0, 10 e 11 passam por um intervalo com os dois extremos inclusivos.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
nota = 0
print(nota, 0 <= nota <= 10)
nota = 10
print(nota, 0 <= nota <= 10)
nota = 11
print(nota, 0 <= nota <= 10)
````

A comparação encadeada exige que nota seja pelo menos 0 e no máximo 10.

## E3 · Resolva

Uma oficina admite quem tem cadastro e tem pelo menos 16 anos. Demonstre três casos, nesta ordem: cadastro e 15 anos, cadastro e 16 anos, sem cadastro e 18 anos. O segundo caso testa o limite inclusivo.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
cadastro = True
idade = 15
print(cadastro and idade >= 16)

cadastro = True
idade = 16
print(cadastro and idade >= 16)

cadastro = False
idade = 18
print(cadastro and idade >= 16)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

```text
True
False
```

Cada situação precisa passar pelas duas verificações. Uma pessoa cadastrada de exatamente 16 anos passa por causa da igualdade em >=.
