# While e parada

Que mudança faz uma contagem iniciada em 3 acabar? Nesta aula você vai prever a parada e usar uma entrada para sinalizar o fim de um lote. Retome [For e acumuladores](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/for-e-acumuladores) e [Entrada e erros](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/entrada-e-erros) se precisar.

Em `while faltam > 0:`, Python testa a condição antes de cada passagem. Começando em 3 e subtraindo 1 a cada vez, `faltam` chega a 0. A condição falha e o corpo não roda novamente. Se a atualização falta ou move o estado na direção errada, o laço pode nunca parar.

Uma sentinela é um valor separado para indicar o fim da leitura. Nesta aula será 0. Dentro de `while True`, `break` encerra imediatamente o laço. Confira a sentinela antes de incrementar a contagem ou a soma. Se precisar parar uma execução no terminal, use Ctrl+C. No caderno, use o controle **Interromper kernel**.

O que deve acontecer quando a primeira entrada já é a sentinela? Nas próximas aulas você vai processar sequências e distinguir valores ausentes de zeros legítimos.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-while-e-parada-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja os valores impressos antes da condição falhar. Conte de 3 até 1 com while.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
faltam = 3
while faltam > 0:
    print(faltam)
    faltam = faltam - 1
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
3
2
1
```

O 0 é alcançado depois da última atualização. O próximo teste é falso e impede que 0 seja impresso.

## E3 · Resolva

Leia um inteiro por vez. Digite `3`, `2` e `0`, confirmando cada linha. Mostre a quantidade de entradas anteriores ao zero e a soma delas, sem contar a sentinela. Repita com 0 como primeira entrada.

Entradas de teste, uma por linha:

```text
3
2
0
```


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
_entradas = iter(['3', '2', '0'])
def input(prompt=''):
    print(prompt, end='')
    return next(_entradas)
````

````python
quantidade = 0
total = 0
while True:
    valor = int(input())
    if valor == 0:
        break
    quantidade = quantidade + 1
    total = total + valor
print("Quantidade:", quantidade)
print("Total:", total)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
del input
````


Saída esperada

```text

Total: 5
```

A checagem antes da atualização impede que zero conte como entrada. Com 0 no primeiro pedido, ambos os acumuladores permanecem em 0. Forneça inteiros até o zero para encerrar.
