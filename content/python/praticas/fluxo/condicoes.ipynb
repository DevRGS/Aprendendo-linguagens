# Condições e faixas

Em um boletim, qual mensagem deve aparecer para quem tirou exatamente 7? Você vai decidir qual bloco executa a partir de uma comparação. Revise [Comparações e lógica](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/comparacoes) se precisar.

Os dois pontos em `if nota >= 7:` abrem um bloco. Desloque as linhas do bloco quatro espaços para a direita. `elif` verifica outra condição apenas quando o `if` e os `elif` anteriores foram falsos. `else` cobre o que ainda resta. Uma cadeia executa o primeiro bloco aplicável. Duas condições podem ser verdadeiras para a mesma nota, então experimente localizar a primeira para cada valor.

Para classificar apenas notas de 0 a 10, descarte inválidas antes das faixas. Depois de descartar essas notas e as aprovadas, `nota >= 5` identifica a recuperação. Pergunte qual bloco fica para uma nota válida de 4.9. Em seguida você vai percorrer vários valores com um laço.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-condicoes-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o rótulo da nota 7 quando a aprovação exige pelo menos 7. Rode e identifique as linhas dentro do bloco.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
nota = 7
if nota >= 7:
    print("aprovada")
else:
    print("abaixo de 7")
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada


O limite inclui 7, portanto só o bloco do if é executado.


## E2 · Altere ou depure

A nota 6 deve ir para recuperação. Compare a primeira cadeia, intencionalmente mal ordenada, com a correção. Preveja qual condição esconde a recuperação.


````python
_pasta = _trabalho / 'e2'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
nota = 6
if nota < 7:
    print("abaixo de 7")
elif nota >= 5 and nota < 7:
    print("recuperação")

if nota >= 7:
    print("aprovada")
elif nota >= 5:
    print("recuperação")
else:
    print("reprovada")
````

No primeiro trecho, a primeira condição já aceita 6. No segundo, a faixa de recuperação fica depois da aprovação e antes do caso restante.
