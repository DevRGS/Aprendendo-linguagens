# Pipeline de arquivos

Encadeie CSV, cálculo e gravação JSON. Preveja o que acontece ao repetir a operação com a mesma entrada. Retome [CSV](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/csv), [JSON](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/json) e [imports](https://devrgs.github.io/Aprendendo-linguagens/#/python/projetos/modulos-e-imports).

Num pipeline a saída de uma etapa alimenta a próxima. A leitura recebe um caminho e devolve uma lista de inteiros. O cálculo recebe a lista e devolve um dicionário. A gravação recebe o resultado e um caminho de saída. Configure origens e destinos por argumentos. Use modo `w` para gerar um único relatório da entrada a cada execução.

No caderno, use os auxiliares da preparação. Reinicie o kernel para reler um auxiliar já importado que tenha editado. No terminal, execute a partir da pasta do exercício para que os caminhos curtos apontem aos dados. Depois de comparar duas execuções, considere o que deve conter um resumo sem registros. Em seguida, veja [instâncias](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/instancias).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-pipeline-de-arquivos-', dir=_raiz))
import numpy
````

## E1 · Preveja e execute

Encadeie as duas funções auxiliares para ler `sessoes.csv` com 3 e 7 minutos. Grave `resumo.json` e mostre o total.

Se editar um auxiliar no painel de arquivos, salve a edição. Execute novamente a preparação desta etapa e a célula do exercício. A preparação inicial do caderno cria outra pasta.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'ler_sessoes.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('import csv\n\ndef ler_minutos(caminho):\n    with caminho.open(encoding="utf-8", newline="") as arquivo:\n        return [int(linha["minutos"]) for linha in csv.DictReader(arquivo)]\n', encoding='utf-8')
_arquivo = _pasta / 'resumir_sessoes.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('def resumir(minutos):\n    return {"total": sum(minutos), "quantidade": len(minutos)}\n', encoding='utf-8')
_arquivo = _pasta / 'sessoes.csv'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('id,minutos\na,3\nb,7\n', encoding='utf-8')
_refresh_imports()
````

````python
import json
from pathlib import Path
from ler_sessoes import ler_minutos
from resumir_sessoes import resumir

def executar(origem, destino):
    minutos = ler_minutos(origem)
    resultado = resumir(minutos)
    with destino.open("w", encoding="utf-8") as arquivo:
        json.dump(resultado, arquivo)
    return resultado

resultado = executar(Path("sessoes.csv"), Path("resumo.json"))
print(resultado["total"])
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
10
```

Ler converte as durações. Resumir calcula o total e executar grava o resultado. Abra resumo.json para ver os dois campos.
