# Ponto de entrada e argumentos

Faça um relatório das primeiras tarefas até um limite. Preveja a quantidade para `--limite 2`. Retome [funções](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno) e [imports](https://devrgs.github.io/Aprendendo-linguagens/#/python/projetos/modulos-e-imports).

`main(args)` coordena a execução. O módulo auxiliar define main. Já o ponto de entrada protege a chamada com `if __name__ == "__main__":`. `argparse.ArgumentParser` descreve opções com tipo e valor padrão. Passar uma lista explícita a `parse_args` evita ler os argumentos do kernel. Sem lista, o parser lê os argumentos do processo.

No terminal, use o arquivo `relatorio_cli.py` de cada exercício. No caderno, passe listas explícitas. Em E3, `runpy` executa o ponto de entrada como script para comparar a saída. A atividade restaura os argumentos da sessão em seguida. O que deve acontecer se a opção for omitida? Veja [pipeline](https://devrgs.github.io/Aprendendo-linguagens/#/python/projetos/pipeline-de-arquivos) depois.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-entrada-do-programa-', dir=_raiz))
import numpy
````

## E1 · Preveja e execute

Receba limite 5 pela lista explícita. No terminal da pasta E1, compare com `python3 relatorio_cli.py --limite 5` em Bash ou `py relatorio_cli.py --limite 5` em PowerShell.

Se editar um auxiliar no painel de arquivos, salve a edição. Execute novamente a preparação desta etapa e a célula do exercício. A preparação inicial do caderno cria outra pasta.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'entrada_limite.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('from argparse import ArgumentParser\n\ndef main(args=None):\n    parser = ArgumentParser()\n    parser.add_argument("--limite", type=int, default=3)\n    opcoes = parser.parse_args(args)\n    print(f"limite: {opcoes.limite}")\n', encoding='utf-8')
_arquivo = _pasta / 'relatorio_cli.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('from entrada_limite import main\n\nif __name__ == "__main__":\n    main()\n', encoding='utf-8')
_refresh_imports()
````

Compare análise válida, pedido de ajuda e opção não numérica. A ajuda e o erro encerrariam o processo e são capturados aqui. Determine qual usa a saída de diagnóstico.


````python
_pasta = _trabalho / 'e2'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
from argparse import ArgumentParser
from contextlib import redirect_stdout, redirect_stderr
from io import StringIO

parser = ArgumentParser(prog="relatorio")
parser.add_argument("--limite", type=int, default=3)
print(parser.parse_args(["--limite", "5"]).limite)

for rotulo, args in [("ajuda", ["--help"]), ("erro", ["--limite", "dois"])]:
    saida, diagnostico = StringIO(), StringIO()
    with redirect_stdout(saida), redirect_stderr(diagnostico):
        try:
            parser.parse_args(args)
        except SystemExit as fim:
            codigo = fim.code
    print(rotulo, codigo, bool(saida.getvalue()), bool(diagnostico.getvalue()))
````

## E3 · Resolva

Faça `main(["--limite", "2"])` e a chamada `relatorio_cli.py --limite 2` emitirem o mesmo relatório com duas tarefas. Use 3 como padrão e 0 como seleção vazia. Na pasta E3, experimente `python3 relatorio_cli.py --limite 2` em Bash ou `py relatorio_cli.py --limite 2` em PowerShell.

Se editar um auxiliar no painel de arquivos, salve a edição. Execute novamente a preparação desta etapa e a célula do exercício. A preparação inicial do caderno cria outra pasta.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'entrada_relatorio.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('from argparse import ArgumentParser\n\ndef main(args=None):\n    parser = ArgumentParser(prog="relatorio")\n    parser.add_argument("--limite", type=int, default=3)\n    opcoes = parser.parse_args(args)\n    if opcoes.limite < 0:\n        parser.error("--limite deve ser >= 0")\n\n    tarefas = ["ler", "revisar", "salvar"][:opcoes.limite]\n    print(f"tarefas={len(tarefas)}")\n    if tarefas:\n        print(", ".join(tarefas))\n', encoding='utf-8')
_arquivo = _pasta / 'relatorio_cli.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('from entrada_relatorio import main\n\nif __name__ == "__main__":\n    main()\n', encoding='utf-8')
_refresh_imports()
````

````python
import runpy
import sys
from contextlib import redirect_stdout
from io import StringIO
from entrada_relatorio import main

no_caderno = StringIO()
with redirect_stdout(no_caderno):
    main(["--limite", "2"])

na_cli = StringIO()
argumentos_anteriores = sys.argv
try:
    sys.argv = ["relatorio_cli.py", "--limite", "2"]
    with redirect_stdout(na_cli):
        runpy.run_path("relatorio_cli.py", run_name="__main__")
finally:
    sys.argv = argumentos_anteriores

print(no_caderno.getvalue(), end="")
print("mesmo relatório:", no_caderno.getvalue() == na_cli.getvalue())
main(["--limite", "0"])
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
tarefas=2
ler, revisar
mesmo relatório: True
tarefas=0
```

main recebe a lista do caderno. O ponto de entrada chama main sem lista e o parser usa sys.argv. Zero seleciona uma fatia vazia. Valores negativos são rejeitados para que não sejam tratados como índices de fim negativos.
