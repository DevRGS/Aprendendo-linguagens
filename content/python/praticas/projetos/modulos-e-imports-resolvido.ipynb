# Módulos e imports locais

Faça um segundo arquivo usar a conversão escrita no primeiro. Preveja `medidas.minutos(1, 5)` e retome [funções](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno).

Um módulo é, aqui, um arquivo Python com definições reutilizáveis. `import medidas` disponibiliza o objeto módulo. Seu espaço de nomes contém `minutos`, acessível como `medidas.minutos`. O Python procura módulos pelo nome. No terminal, a pasta do script entra na busca no uso normal. No caderno, a preparação disponibiliza as pastas de exercício para importação.

Defina funções no auxiliar e chame-as no consumidor. Um kernel pode reutilizar um módulo importado sem reler o arquivo. Após editar um auxiliar no navegador, salve e execute novamente a preparação desta etapa e a célula que importa o código. Isso reutiliza a pasta desta atividade. No terminal, rodar o script novamente inicia outro processo. Por que o nome do seu arquivo pode mudar um import? Em seguida, veja [ambientes e pacotes](https://devrgs.github.io/Aprendendo-linguagens/#/python/projetos/ambiente-e-pacotes).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-modulos-e-imports-', dir=_raiz))
import numpy
````

## E1 · Preveja e execute

Confira `medidas.py` junto da atividade. Importe-o e use a função com 1 hora e 5 minutos.

Se editar um auxiliar no painel de arquivos, salve a edição. Execute novamente a preparação desta etapa e a célula do exercício. A preparação inicial do caderno cria outra pasta.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'medidas.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('def minutos(horas, restantes):\n    return horas * 60 + restantes\n', encoding='utf-8')
_refresh_imports()
````

## E3 · Resolva

Escreva `minutos(h, m)` e `segundos(m)` no auxiliar `conversoes.py`. Importe as duas funções em um segundo arquivo, ou na célula do caderno, e mostre 65 e 3900 com (1, 5). A importação deve ser silenciosa, e zero minutos devem resultar em 0 segundos.

Se editar um auxiliar no painel de arquivos, salve a edição. Execute novamente a preparação desta etapa e a célula do exercício. A preparação inicial do caderno cria outra pasta.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'conversoes.py'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('def minutos(horas, restantes):\n    return horas * 60 + restantes\n\n\ndef segundos(minutos):\n    return minutos * 60\n', encoding='utf-8')
_refresh_imports()
````

````python
from contextlib import redirect_stdout
from io import StringIO

durante_import = StringIO()
with redirect_stdout(durante_import):
    from conversoes import minutos, segundos

assert durante_import.getvalue() == ""
duracao = minutos(1, 5)
print(duracao, segundos(duracao))
print(segundos(minutos(0, 0)))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````
