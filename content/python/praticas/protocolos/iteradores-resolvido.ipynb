# Iteráveis, iteradores e esgotamento

Leia dois valores, um por vez. Antes da terceira chamada, preveja se o cursor retorna ao início. Retome [laços](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/for-e-acumuladores) e [listas](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/listas).

Um iterável pode fornecer um iterador por `iter(objeto)`. Um iterador guarda a posição, e `next(iterador)` solicita o próximo valor. Quando acaba, a chamada sem padrão levanta `StopIteration`. Um laço for usa esse sinal para terminar. Já `next(iterador, padrão)` devolve o padrão no esgotamento. Uma lista normalmente fornece um iterador novo a cada percurso. Pedir `iter` ao próprio iterador devolve o mesmo objeto.

Use cada bloco na célula ou pasta do próprio exercício. O esgotamento de E1 é intencional. Que padrão você escolheria se zero fosse um valor válido? Em seguida, veja [geradores](https://devrgs.github.io/Aprendendo-linguagens/#/python/protocolos/geradores).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-iteradores-', dir=_raiz))
````

## E1 · Preveja e execute

Consuma [2, 5] com duas chamadas. Faça uma terceira chamada intencionalmente e capture apenas StopIteration.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
cursor = iter([2, 5])
print(next(cursor))
print(next(cursor))
try:
    next(cursor)
except StopIteration:
    print("fim do iterador")
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
2
5
fim do iterador
```

O iterador lembra onde parou. O terceiro next sinaliza esgotamento, e o tratamento mostra isso em uma linha.

## E3 · Resolva

Obtenha apenas o primeiro inteiro não negativo em [None, -1, 0, 4]. Use None como padrão para uma entrada sem válidos, inclusive a sequência vazia.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def primeiro_valido(registros):
    validos = (n for n in registros if type(n) is int and n >= 0)
    return next(validos, None)

print(primeiro_valido([None, -1, 0, 4]))
print(primeiro_valido([None, -2]))
print(primeiro_valido([]))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
0
None
None
```

A expressão com parênteses produz os candidatos sob demanda. next só pede um. type(n) is int restringe a inteiros, excluindo bool, e a segunda condição aceita 0. None identifica esgotamento sem ser uma resposta válida.
