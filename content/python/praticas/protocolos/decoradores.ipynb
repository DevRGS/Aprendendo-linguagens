# Decoradores e encaminhamento de chamadas

Envolva uma função de cálculo com um registro de chamadas e verifique se o valor calculado é preservado. Retome [funções como valores](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/composicao) e [retorno](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno).

Uma função de ordem superior recebe ou devolve funções. No decorador, a função interna guarda acesso ao argumento da função externa. Isso é uma closure. Essa função interna, chamada wrapper, roda no lugar da original. A notação `@decorar` equivale a definir a função e aplicar `nome = decorar(nome)`. `functools.wraps` copia metadados, como `__name__`. Encaminhar `*args` e `**kwargs` e usar `return` preserva a chamada e o resultado.

Cada bloco define seu próprio decorador na célula ou no kit. O erro do segundo exercício é intencional. O que permanece no objeto função em E3 depois de chamar duas vezes? Em seguida, veja [gerentes de contexto](https://devrgs.github.io/Aprendendo-linguagens/#/python/protocolos/gerentes-de-contexto).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-decoradores-', dir=_raiz))
````

## E1 · Preveja e execute

Envolva uma função dobro, registre uma chamada e confira se o registro aparece antes ou depois do resultado.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def registrar(funcao):
    def wrapper(valor):
        print("chamada registrada")
        return funcao(valor)
    return wrapper

@registrar
def dobro(n):
    return n * 2

print(dobro(3))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
chamada registrada
6
```

dobro foi substituída pelo wrapper. Ele registra e devolve o cálculo da função que guardou.


## E2 · Altere ou depure

Investigue o wrapper que perde o retorno e não aceita argumento nomeado. Capture o TypeError intencional e compare com o wrapper reparado.


````python
_pasta = _trabalho / 'e2'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def defeituoso(funcao):
    def wrapper(*args):
        funcao(*args)
    return wrapper

def corrigido(funcao):
    def wrapper(*args, **kwargs):
        return funcao(*args, **kwargs)
    return wrapper

def somar(a, b=0):
    return a + b

ruim = defeituoso(somar)
print(ruim(2, 3))
try:
    ruim(2, b=3)
except TypeError:
    print("argumento nomeado rejeitado")

bom = corrigido(somar)
print(bom(2, b=3))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
None
argumento nomeado rejeitado
5
```

Sem return, o wrapper devolve None. Sem kwargs, a chamada nomeada falha na entrada do wrapper. A correção encaminha os dois grupos de argumentos e devolve o resultado.
