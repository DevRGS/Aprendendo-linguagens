# Geradores sob demanda

Solicite só o primeiro quadrado e observe quantas multiplicações acontecem. Retome [iteradores](https://devrgs.github.io/Aprendendo-linguagens/#/python/protocolos/iteradores).

Uma função com `yield` cria um gerador ao ser chamada. O corpo começa quando pedimos um valor. `yield` entrega o valor e suspende o corpo, que continua na próxima solicitação. Como iterador, o gerador esgota. Consumir itens sob demanda permite processar dados grandes sem guardar a sequência inteira. Transformá-los em `list` materializa todos os itens.

Use o CSV ou texto fornecido apenas onde indicado no enunciado. O arquivo de E3 fica aberto durante o consumo da função geradora. A atividade consome até o final dentro do cálculo. O que mudaria se o consumidor pedisse só um valor? Depois, veja [decoradores](https://devrgs.github.io/Aprendendo-linguagens/#/python/protocolos/decoradores).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-geradores-', dir=_raiz))
````

## E1 · Preveja e execute

Gere os quadrados dos inteiros 0, 1 e 2. Antes do laço, diga em que momento a função geradora calculará cada item.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def quadrados(limite):
    for n in range(limite):
        yield n * n

for valor in quadrados(3):
    print(valor)
````

````python
def quadrados(limite):
    for n in range(limite):
        print(f"calculando {n}")
        yield n * n

cursor = quadrados(3)
print("gerador criado")
print("primeiro:", next(cursor))
````
