# Testes unitários com unittest

Verifique a conversão de uma hora e cinco minutos. Antes de chamar sua função, calcule de forma independente o resultado esperado. Retome [retorno de funções](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno) e [exceções](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/excecoes).

Um teste prepara uma entrada, executa a ação e compara o resultado com uma previsão. Use `unittest.TestCase` e nomes de métodos que comecem com `test_`. `assertEqual` compara o valor obtido ao esperado. `assertRaises` confere uma exceção prometida pelo contrato. O `TestLoader` carrega os métodos da classe para uma suíte e `TextTestRunner` executa a suíte e devolve o relatório.

Use a suíte explícita de cada exercício, tanto no kit como no caderno. Assim, os argumentos do kernel não entram na execução. Aqui o runner grava o relatório em memória para evitar que o tempo variável faça parte da saída esperada. Para inspecioná-lo, acrescente `print(registro.getvalue())`. Qual borda detecta um retorno que descarta minutos restantes? Depois, veja [depuração](https://devrgs.github.io/Aprendendo-linguagens/#/python/qualidade/depuracao).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-testes-unitarios-', dir=_raiz))
````

## E1 · Preveja e execute

Escreva o teste de minutos(1, 5) com esperado 65. Carregue a classe de teste, execute um caso e confira a contagem.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import io
import unittest

def minutos(horas, restantes):
    return horas * 60 + restantes

class TestesMinutos(unittest.TestCase):
    def test_hora_e_minutos(self):
        self.assertEqual(minutos(1, 5), 65)


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestesMinutos)
registro = io.StringIO()
resultado = unittest.TextTestRunner(stream=registro).run(suite)

print("casos:", resultado.testsRun, "sucesso:", resultado.wasSuccessful())
````

## E3 · Resolva

Crie três testes de minutos(horas, restantes), para (1, 5), (0, 0) e (0, 60). O domínio tem horas inteiras não negativas e minutos restantes entre 0 e 59. Caso inválido deve levantar ValueError, verificado com assertRaises.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import io
import unittest

def minutos(horas, restantes):
    if type(horas) is not int or type(restantes) is not int:
        raise ValueError("horas e minutos devem ser inteiros")
    if horas < 0 or not 0 <= restantes <= 59:
        raise ValueError("horas ou minutos fora do intervalo")
    return horas * 60 + restantes

class TestesMinutos(unittest.TestCase):
    def test_conversao(self):
        self.assertEqual(minutos(1, 5), 65)

    def test_limite_inferior(self):
        self.assertEqual(minutos(0, 0), 0)

    def test_minutos_fora_do_intervalo(self):
        with self.assertRaises(ValueError):
            minutos(0, 60)


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestesMinutos)
registro = io.StringIO()
resultado = unittest.TextTestRunner(stream=registro).run(suite)

print("casos:", resultado.testsRun, "sucesso:", resultado.wasSuccessful())
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
casos: 3 sucesso: True
```

Zero horas e zero minutos formam a borda inferior válida. Sessenta minutos restantes violam o contrato. O teste passa quando a chamada levanta a classe esperada. Confira sempre testsRun para detectar a suíte vazia.
