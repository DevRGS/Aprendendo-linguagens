# Dicionários

Como guardar uma medida por ID e substituir só a medida de a? Retome [Tuplas e desempacotamento](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/tuplas-e-desempacotamento) para rever registros em pares.

Um dicionário associa chaves a valores. Em `medidas = {"a": 2}`, a chave é o texto a e o valor é 2. `medidas["a"]` devolve 2. Se a chave não existir, esse acesso obrigatório gera `KeyError`. `medidas.get("b", 0)` devolve 0 como padrão se não houver b. Se zero for um valor registrado, use `"b" in medidas` para testar presença.

`medidas["a"] = 3` substitui apenas a associação de a. `for chave, valor in medidas.items():` entrega cada par para desempacotamento. Antes de avançar, explique quando usar padrão 0 faz sentido, como em uma contagem que ainda não viu a cor. Na próxima aula você vai verificar chaves comuns sem guardar valores.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-dicionarios-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja o valor de a antes e depois da atualização. Leia e exiba o campo nas duas etapas.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
medidas = {"a": 2}
print(medidas["a"])
medidas["a"] = 3
print(medidas["a"])
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

3
```

A segunda atribuição usa a mesma chave e substitui o valor associado.

## E3 · Resolva

Conte as ocorrências de cada cor em [azul, azul, verde]. Imprima uma linha por cor na ordem da primeira aparição. Confira também se uma entrada vazia produz zero associações.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
cores = ["azul", "azul", "verde"]
frequencias = {}
for cor in cores:
    frequencias[cor] = frequencias.get(cor, 0) + 1
for cor, quantidade in frequencias.items():
    print(cor, quantidade)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
azul 2
verde 1
```

Cada ocorrência acrescenta 1 à frequência anterior, que começa em 0 para uma chave nova. items percorre na ordem de inserção. Sem ocorrências, o mapa fica vazio e o segundo laço não imprime linhas.
