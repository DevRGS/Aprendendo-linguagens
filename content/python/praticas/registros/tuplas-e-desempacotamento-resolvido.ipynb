# Tuplas e desempacotamento

Como manter o ID de uma leitura ligado ao seu valor? Retome [Listas e cópias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/listas). Aqui cada registro tem dois campos em ordem fixa.

A tupla `("a", 3)` guarda texto e número em um mesmo registro. A tupla é imutável: a posição 0 continua com o mesmo item. Em `identificador, valor = registro`, o desempacotamento liga cada nome ao item da posição correspondente. Para dois nomes, o registro precisa ter dois itens.

`zip(nomes, medidas)` produz pares, um por posição. Por padrão, para assim que uma das entradas termina. Com nomes `["a", "b"]` e medidas `[3]`, só aparece `("a", 3)`. Se todos os nomes precisarem de medida, compare os comprimentos antes de combinar ou use `zip(..., strict=True)` e percorra os resultados para detectar a divergência. Nesta aula usamos listas, cujo tamanho se consulta com `len`.

Explique por que o truncamento pode esconder um registro incompleto. Em [Dicionários](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/dicionarios) você vai usar o ID para localizar um valor.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.

## E3 · Resolva

Combine os nomes A, B e C com as medidas 2, 3 e 4, na mesma ordem. Exiba cada par. Se o tamanho divergir, rejeite o lote com `Tamanhos incompatíveis`, sem imprimir pares parciais. Teste com a última medida removida.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
nomes = ["A", "B", "C"]
medidas = [2, 3, 4]

if len(nomes) != len(medidas):
    print("Tamanhos incompatíveis")
else:
    for nome, medida in zip(nomes, medidas):
        print(nome, medida)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

B 3
C 4
```

Comparar os tamanhos antes da iteração impede que parte do lote seja mostrada. Com duas medidas, só sai a mensagem de incompatibilidade. Duas listas vazias têm tamanhos compatíveis e produzem zero pares.
