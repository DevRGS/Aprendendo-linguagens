# Conjuntos

Dois lotes podem repetir um ID dentro do próprio lote. Como contar IDs distintos? Retome [Listas e cópias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/listas) e [Comparações e lógica](https://devrgs.github.io/Aprendendo-linguagens/#/python/fluxo/comparacoes).

`set([2, 2, 4])` guarda uma vez cada ID. `2 in ids` testa pertinência. A interseção `a & b` contém os itens presentes em ambos. A diferença `a - b` contém os itens de a que estão fora de b. O conjunto não oferece uma ordem de exibição para o relatório. `sorted(ids)` devolve uma lista ordenada dos IDs comparáveis.

`set("ab")` percorre os caracteres do texto. Já `{"ab"}` contém o texto inteiro como um elemento. Para escrever um conjunto vazio, use `set()`, pois `{}` cria um dicionário. Antes de seguir para [Resumo de lote](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/resumo-de-lote), explique com um exemplo por que a diferença depende da ordem dos lotes.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-conjuntos-', dir=_raiz))
````

## E1 · Preveja e execute

Preveja quantos IDs únicos há em [2, 2, 4]. Mostre a lista ordenada e a quantidade.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
ids = set([2, 2, 4])
print(sorted(ids))
print(len(ids))
````
