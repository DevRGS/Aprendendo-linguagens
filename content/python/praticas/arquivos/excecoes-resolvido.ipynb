# Tratar erros com contexto

Preveja a diferença entre abrir um arquivo ausente e converter “dez” em inteiro. Retome [entrada e erros](https://devrgs.github.io/Aprendendo-linguagens/#/python/inicio/entrada-e-erros) e [leitura de arquivos](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/caminhos-e-texto).

Para interromper e acrescentar contexto, `raise ValueError(mensagem) from erro` preserva a causa original.

Compare um arquivo ausente e outro com a segunda linha inválida. Mostre o número da linha e confira a causa original.

````python
from pathlib import Path

def ler_total(caminho):
    total = 0
    with caminho.open(encoding="utf-8") as arquivo:
        for numero, linha in enumerate(arquivo, start=1):
            try:
                total += int(linha)
            except ValueError as erro:
                raise ValueError(f"linha {numero}: inteiro inválido") from erro
    return total

for nome in ["sem-medidas.txt", "medidas.txt"]:
    try:
        print(ler_total(Path(nome)))
    except FileNotFoundError:
        print("arquivo ausente")
    except ValueError as erro:
        print(erro)
        print(f"causa: {type(erro.__cause__).__name__}")
````

## E3 · Resolva

Leia `lote.txt` com `3`, `erro` e `0` em três linhas. Pule somente conversões inválidas. Mostre dois válidos, uma rejeição e soma 3. Um arquivo vazio deve dar três zeros.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'lote.txt'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('3\nerro\n0\n', encoding='utf-8')
_refresh_imports()
````

````python
from pathlib import Path

def resumir_lote(caminho):
    validos = rejeitados = soma = 0
    with caminho.open(encoding="utf-8") as arquivo:
        for linha in arquivo:
            try:
                numero = int(linha)
            except ValueError:
                rejeitados += 1
                continue
            validos += 1
            soma += numero
    return validos, rejeitados, soma

validos, rejeitados, soma = resumir_lote(Path("lote.txt"))
print(f"válidos={validos}, rejeitados={rejeitados}, soma={soma}")
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

```

Toda conversão bem-sucedida conta, inclusive zero. Linha em branco é rejeitada por int. No arquivo vazio o laço não roda. Erro ao abrir se propaga.
