# Gravar e carregar JSON

Você vai guardar um campo sem valor. Preveja como ele aparece no arquivo. Retome [dicionários](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/dicionarios) e [UTF-8](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/caminhos-e-texto).

JSON é texto. Para os dados desta aula, dicionários correspondem a objetos JSON, listas a arrays, e `None` a `null`. `json.dump(dados, arquivo)` grava o texto e `json.load(arquivo)` interpreta o texto como um objeto Python. `dumps` e `loads` fazem esse percurso com uma string.

Abra os arquivos com UTF-8 no caderno e no terminal. `ensure_ascii=False` deixa os acentos legíveis no texto gravado. Um campo presente com valor nulo pode exigir um tratamento diferente de uma chave ausente. Qual teste distinguiria os dois? Depois veja [exceções](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/excecoes).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-json-', dir=_raiz))
````

## E1 · Preveja e execute

Grave um dicionário com uma lista de temas e um campo fim igual a None. Leia o texto e compare o objeto carregado ao original.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import json
from pathlib import Path

original = {"temas": ["ação", "dados"], "fim": None}
caminho = Path("temas.json")
with caminho.open("w", encoding="utf-8") as arquivo:
    json.dump(original, arquivo, ensure_ascii=False)
print(caminho.read_text(encoding="utf-8"))

with caminho.open(encoding="utf-8") as arquivo:
    recuperado = json.load(arquivo)
print(recuperado == original)
````
