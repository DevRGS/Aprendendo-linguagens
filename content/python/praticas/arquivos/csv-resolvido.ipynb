# Ler e escrever CSV

Some durações de dois registros. Antes de rodar, preveja se `minutos` chega à leitura como número ou texto. Retome [arquivos de texto](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/caminhos-e-texto) e [dicionários](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/dicionarios).

Converta `linha["minutos"]` com `int` antes de somar. Abra CSV com `encoding="utf-8", newline=""` para que a biblioteca trate as quebras de linha, inclusive dentro de campos entre aspas.

A preparação do caderno cria as entradas de cada exercício. No terminal, localize-as na pasta do exercício do kit. Depois de reler a tabela filtrada, pense em como guardar números sem convertê-los a cada leitura e siga para [JSON](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/json).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.

## E1 · Preveja e execute

Leia `sessoes.csv` com cabeçalho `id,minutos` e registros de 3 e 7 minutos. Mostre o tipo de cada campo minutos antes da conversão e o total 10.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'sessoes.csv'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('id,minutos\na,3\nb,7\n', encoding='utf-8')
_refresh_imports()
````

````python
from pathlib import Path
from csv import DictReader

with Path("sessoes.csv").open(encoding="utf-8", newline="") as arquivo:
    total = 0
    for linha in DictReader(arquivo):
        print(type(linha["minutos"]).__name__)
        total += int(linha["minutos"])
print(total)
````

## E3 · Resolva

Filtre `sessoes.csv` por minutos ≥ 5, grave `selecionadas.csv` com cabeçalho `id,minutos` e releia. Mostre pares com ID como texto e minutos como inteiro. Sem linhas selecionadas, preserve o cabeçalho.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_arquivo = _pasta / 'sessoes.csv'
_arquivo.parent.mkdir(parents=True, exist_ok=True)
if not _arquivo.exists():
    _arquivo.write_text('id,minutos\n01,3\n02,7\n03,5\n', encoding='utf-8')
_refresh_imports()
````

````python
from pathlib import Path
from csv import DictReader, DictWriter

destino = Path("selecionadas.csv")
with Path("sessoes.csv").open(encoding="utf-8", newline="") as entrada:
    with destino.open("w", encoding="utf-8", newline="") as saida:
        escritor = DictWriter(saida, fieldnames=["id", "minutos"])
        escritor.writeheader()
        for linha in DictReader(entrada):
            if int(linha["minutos"]) >= 5:
                escritor.writerow(linha)

with destino.open(encoding="utf-8", newline="") as saida:
    for linha in DictReader(saida):
        print((linha["id"], int(linha["minutos"])))
````
