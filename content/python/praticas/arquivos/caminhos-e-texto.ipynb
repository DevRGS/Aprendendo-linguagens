O diretório de trabalho é a pasta a partir da qual caminhos relativos são interpretados. `from pathlib import Path` torna o nome `Path` disponível. `Path("mensagem.txt")` representa um caminho relativo, `Path.cwd()` identifica a pasta atual e `resolve()` obtém o caminho absoluto.

UTF-8 permite converter bytes em texto com acentos como “ação”. Use o modo `r` para ler, `w` para escrever truncando o conteúdo existente ou `x` para criar um arquivo novo sem sobrescrever. `with` fecha o arquivo ao sair do bloco.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-caminhos-e-texto-', dir=_raiz))
````

## E1 · Preveja e execute

Numa pasta temporária da atividade, grave e releia “ação”. Preveja quais caracteres aparecem.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory(dir=Path.cwd()) as pasta:
    caminho = Path(pasta) / "mensagem.txt"
    with caminho.open("x", encoding="utf-8") as arquivo:
        arquivo.write("ação")
    with caminho.open("r", encoding="utf-8") as arquivo:
        print(arquivo.read())
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

```

O modo x cria o arquivo. A escrita termina antes da leitura, que recupera os caracteres em UTF-8. A pasta temporária é nova a cada execução.
