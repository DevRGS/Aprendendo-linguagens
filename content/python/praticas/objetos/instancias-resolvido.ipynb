# Instâncias e estado

Crie dois contadores. Antes de incrementar um deles, preveja o valor que ficará no outro. Retome [funções e argumentos](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/parametros-e-retorno).

Uma classe define como criar e usar um tipo de objeto. Cada chamada a `Contador()` cria uma instância. O método `__init__` inicializa essa instância. Em `a.incrementar()`, Python passa o objeto `a` como primeiro argumento do método, chamado `self`. Atribuir `self.valor` guarda estado naquela instância.

Execute cada bloco em sua pasta do kit ou na célula correspondente do caderno. Que lugar dá uma lista própria a cada contador? Na próxima aula, [dataclasses](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/dataclasses) gera parte da inicialização.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-instancias-', dir=_raiz))
````

## E1 · Preveja e execute

Crie e exiba dois contadores, incremente apenas o primeiro e observe os dois estados finais.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
class Contador:
    def __init__(self):
        self.valor = 0

    def incrementar(self):
        self.valor += 1


a, b = Contador(), Contador()
print(a.valor, b.valor)
a.incrementar()
print(a.valor, b.valor)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
0 0
1 0
```

self aponta para a instância que recebeu a chamada. O método mudou apenas a.valor.

## E3 · Resolva

Implemente Acumulador com total inicial 0 e um método adicionar. Adicione 2 e 5 à primeira instância e exiba 7. Uma instância nova ou que recebeu apenas 0 deve expor total 0.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
class Acumulador:
    def __init__(self):
        self.total = 0

    def adicionar(self, valor):
        self.total += valor


acumulador = Acumulador()
acumulador.adicionar(2)
acumulador.adicionar(5)
print(acumulador.total)

vazio = Acumulador()
vazio.adicionar(0)
print(vazio.total)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
7
0
```

O total fica na instância. Cada método adiciona ao valor anterior. Acrescentar zero preserva o total e outra instância começa do próprio zero.
