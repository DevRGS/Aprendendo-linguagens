# Composição e estratégias

Compare duas formas de tratar um rótulo, removendo espaços e convertendo para minúsculas ou preservando o texto. Preveja a diferença para `" Azul "`. Retome [métodos de texto](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/texto-e-fatias) e [instâncias](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/instancias).

Combine objetos por composição. Um processador guarda uma estratégia em um atributo e delega o trabalho ao método `transformar(texto)`. O contrato desta aula pede um texto como argumento e devolve um texto. Estratégias diferentes podem ocupar o mesmo papel quando cumprem esse contrato. Essa substituição pelo comportamento é uma forma de polimorfismo.

Rode os blocos com as definições de cada exercício no caderno ou kit. Que mudança ocorreria se a estratégia recebesse uma lista em vez de texto? Depois avalie [quando escolher modelo ou função](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/modelo-ou-funcao).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-composicao-e-polimorfismo-', dir=_raiz))
````

## E1 · Preveja e execute

As duas estratégias têm o mesmo método. Aplique ambas a " Azul " e mostre onde ficaram os espaços.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
class Normalizar:
    def transformar(self, texto):
        return texto.strip().lower()


class Identidade:
    def transformar(self, texto):
        return texto


for estrategia in [Normalizar(), Identidade()]:
    print(repr(estrategia.transformar(" Azul ")))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

' Azul '
```

O chamador usa a mesma expressão em ambos os casos. repr torna visíveis as aspas e os espaços preservados.


## E2 · Altere ou depure

Troque a normalização pela identidade ao chamar rotular. Antes de rodar, diga onde os espaços aparecerão entre os delimitadores.


````python
_pasta = _trabalho / 'e2'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
class Normalizar:
    def transformar(self, texto):
        return texto.strip().lower()


class Identidade:
    def transformar(self, texto):
        return texto


def rotular(texto, estrategia):
    return "[" + estrategia.transformar(texto) + "]"

texto = " Leitura "
print(rotular(texto, Normalizar()))
print(rotular(texto, Identidade()))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
[leitura]
[ Leitura ]
```

O consumidor continua delegando ao mesmo método. Só o resultado do método muda entre as duas chamadas.


## E3 · Resolva

Construa Processador que recebe uma estratégia no inicializador e processa uma lista de rótulos. Teste Normalizar e Identidade com [" Azul ", "VERDE"]. As saídas devem manter ordem e quantidade. Entrada vazia deve produzir lista vazia.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
class Normalizar:
    def transformar(self, texto):
        return texto.strip().lower()


class Identidade:
    def transformar(self, texto):
        return texto


class Processador:
    def __init__(self, estrategia):
        self.estrategia = estrategia

    def processar(self, rotulos):
        return [self.estrategia.transformar(rotulo) for rotulo in rotulos]


rotulos = [" Azul ", "VERDE"]
normalizador = Processador(Normalizar())
identidade = Processador(Identidade())
print(normalizador.processar(rotulos))
print(identidade.processar(rotulos))
print(normalizador.processar([]))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
['azul', 'verde']
[' Azul ', 'VERDE']
[]
```

O processador guarda a estratégia recebida e aplica seu método a cada entrada na ordem. A compreensão cria nova lista e, se a entrada estiver vazia, não chama a estratégia.
