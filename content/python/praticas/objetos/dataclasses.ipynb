# Registros com dataclasses

Modele dois registros com ID e valor. Preveja como a igualdade funciona se os campos são iguais e os objetos foram criados separadamente. Retome [instâncias](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/instancias) e [leitura de CSV](https://devrgs.github.io/Aprendendo-linguagens/#/python/arquivos/csv).

Os campos anotados de uma classe decorada com `@dataclass` definem parâmetros do inicializador gerado. Com as opções padrão, a igualdade de instâncias da mesma classe compara seus campos. Anotações descrevem os tipos, e a validação dos dados vindos de fora ainda deve ser escrita. `field(default_factory=list)` chama `list()` para cada instância criada sem valor para aquele campo.

A entrada E3 está na pasta do exercício no kit e é criada na preparação do caderno. Qual consequência tem converter um ID com zeros à esquerda? Depois, veja [composição](https://devrgs.github.io/Aprendendo-linguagens/#/python/objetos/composicao-e-polimorfismo).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.

````python
from dataclasses import dataclass, field

@dataclass
class Medicao:
    id: str
    marcas: list[str] = field(default_factory=list)

a = Medicao("a")
b = Medicao("b")
a.marcas.append("revista")
print(a.marcas, b.marcas)
````
