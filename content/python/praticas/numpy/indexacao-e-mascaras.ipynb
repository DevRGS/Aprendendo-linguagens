# Indexação, máscaras e cópias

Você quer manter somente viagens com distância positiva. Preveja os IDs que devem sobreviver se a segunda linha trouxer uma distância negativa.

Use NumPy e retome [Arrays](https://devrgs.github.io/Aprendendo-linguagens/#/python/numpy/arrays) se precisar rever a forma de uma coluna. Cada atividade cria os próprios dados.

Uma comparação com um array produz uma **máscara booleana**, com um verdadeiro ou falso para cada elemento comparado. Se a máscara veio de uma coluna, pode ser usada no primeiro eixo da tabela para selecionar linhas inteiras. Fatias básicas normalmente compartilham os dados com o array de origem, como uma janela para os mesmos valores (**view**). A indexação booleana devolve uma cópia, e `copy()` permite pedir explicitamente uma cópia de uma fatia.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-indexacao-e-mascaras-', dir=_raiz))
import numpy
````

## E1 · Preveja e execute

Em `[2, -1, 5]`, marque os valores positivos antes de executar. Imprima primeiro a máscara e depois a seleção.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````
