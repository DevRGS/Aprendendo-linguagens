# Arrays, tipos e dimensões

Imagine três viagens medidas por distância e tempo. Quantos números descrevem cada viagem? Agora imagine guardar apenas a primeira viagem para um programa que espera a tabela inteira. Que formato essa seleção precisa ter?

Revise listas e fatias em [Texto e fatias](https://devrgs.github.io/Aprendendo-linguagens/#/python/sequencias/texto-e-fatias). No caderno, o kernel precisa oferecer NumPy. No terminal, instale as dependências do kit e rode os scripts com o mesmo intérprete. Nenhum arquivo de entrada é necessário nesta aula.

`np.array` cria um `ndarray`. Este array guarda elementos de um tipo comum, chamado `dtype`. Escolhemos `int64` nos exemplos para tornar explícito o tipo de inteiro usado. `ndim` informa a quantidade de dimensões. `shape` dá o comprimento em cada dimensão, na ordem. Um array de forma `(3, 2)` tem três linhas e duas colunas. Um vetor de forma `(3,)` tem uma dimensão, mesmo se seus três números vieram de uma coluna.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-arrays-', dir=_raiz))
import numpy
````
