# Produto matricial e pontuação

Uma viagem tem dois valores e você quer uma nota calculada com um peso para cada um. Quantas notas você espera de três viagens?

Use NumPy e retome [Eixos e broadcasting](https://devrgs.github.io/Aprendendo-linguagens/#/python/numpy/eixos-e-broadcasting). As atividades criam os arrays diretamente.

`a * b` multiplica elementos, seguindo as regras de broadcasting. `a @ b` realiza produto matricial. Em `(n, p) @ (p, m)`, cada linha da primeira matriz é combinada com cada coluna da segunda. O comprimento interno `p` deve coincidir. A saída tem forma `(n, m)`. Se o segundo operando for um vetor de forma `(p,)`, o resultado fica com forma `(n,)`: uma soma de produtos por linha.

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-produto-matricial-', dir=_raiz))
import numpy
````

## E3 · Resolva

Obtenha uma pontuação por linha de `[[1, 3], [2, 5], [0, 4]]` com pesos `[0.5, 2.0]`. Faça por laços e por `@`, compare com `np.allclose` e teste lote de uma linha.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
import numpy as np

def pontuar_laco(a, pesos):
    notas = []
    for linha in a:
        soma = 0.0
        for valor, peso in zip(linha, pesos, strict=True):
            soma += valor * peso
        notas.append(soma)
    return notas

a = np.array([[1, 3], [2, 5], [0, 4]])
pesos = np.array([0.5, 2.0])
notas = a @ pesos
print(notas.tolist())
print(np.allclose(pontuar_laco(a, pesos), notas))
assert (a[:1] @ pesos).shape == (1,)
assert np.allclose(pontuar_laco(a[:1], pesos), a[:1] @ pesos)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text

True
```

O laço interno combina cada valor com seu peso e acumula a pontuação. `strict=True` evita truncar silenciosamente uma linha de comprimento incompatível. Para uma linha a saída continua sendo um vetor. Se errou o eixo no marco, volte a [Eixos e broadcasting](#/python/numpy/eixos-e-broadcasting) e repita aqui com pesos [1, 0] e uma linha [7, 2].
