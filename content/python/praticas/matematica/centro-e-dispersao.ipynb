# Centro e dispersão

Média é soma dividida pela quantidade. Mediana é o valor central da lista ordenada ou a média dos dois valores centrais. Amplitude subtrai o menor valor do maior. Desvio padrão mede dispersão na unidade original. Quando toda a população é conhecida, a variância usa n no denominador. A variância amostral usa n−1 para estimar a variância populacional a partir da amostra. O desvio padrão é a raiz quadrada da variância. Revise [funções e contratos](https://devrgs.github.io/Aprendendo-linguagens/#/python/funcoes/contratos-e-tipos).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-centro-e-dispersao-', dir=_raiz))
````

## E1 · Preveja e execute

Calcule média, mediana e amplitude de [2,4,6].


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
from statistics import mean, median
valores = [2,4,6]
print(mean(valores), median(valores), max(valores)-min(valores))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````
