# Frequência, condição e probabilidade

O espaço amostral reúne os resultados possíveis de um experimento. Em um conjunto observado, frequência relativa divide ocorrências pelo número de observações. Para uma proporção condicional, restrinja primeiro o grupo e conte dentro dele. Uma proporção conjunta usa o total e conta só as observações que satisfazem as duas condições. Uma simulação usa números pseudoaleatórios para ilustrar a diferença entre frequência observada e probabilidade teórica. ## E3 · Resolva

Simule 20 lançamentos equiprováveis. Compare com 0.5 e verifique que a mesma semente reproduz a frequência.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
from random import Random
def proporcao(seed, n):
    if n <= 0:
        raise ValueError("n deve ser positivo")
    rng = Random(seed)
    return sum(rng.random() < 0.5 for _ in range(n)) / n

observado = proporcao(3, 20)
print(f"observado={observado:.2f} esperado=0.50")
assert proporcao(3, 20) == observado
try:
    proporcao(3, 0)
except ValueError as erro:
    print(erro)
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````
