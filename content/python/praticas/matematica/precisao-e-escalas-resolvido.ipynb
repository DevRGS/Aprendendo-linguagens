# Precisão, centralização e escalas

Números float representam muitos decimais por aproximação binária. Use isclose para verificar uma diferença aceitável e indique tolerância absoluta perto de zero. Centralizar subtrai a média de cada observação. Normalizar pelo intervalo subtrai o mínimo e divide pela amplitude. Um vetor constante exige uma regra porque a amplitude é zero. Guarde a unidade antes de transformar. Revise [centro e dispersão](https://devrgs.github.io/Aprendendo-linguagens/#/python/matematica/centro-e-dispersao).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.

## E3 · Resolva

Normalize [2,4,8] pela amplitude. Comprove, por tolerância, o valor intermediário. Use zero para coluna constante.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
from math import isclose
def escalar(valores):
    minimo, maximo = min(valores), max(valores)
    amplitude = maximo - minimo
    if amplitude == 0:
        return [0.0 for _ in valores]
    return [(x - minimo) / amplitude for x in valores]

normal = escalar([2,4,8])
print([round(x, 3) for x in normal])
assert isclose(normal[1], 1/3, abs_tol=1e-9)
print(escalar([5,5]))
````
