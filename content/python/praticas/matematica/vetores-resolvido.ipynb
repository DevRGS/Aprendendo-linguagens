# Vetores e soma ponderada

Cada linha de uma tabela pode descrever uma observação por meio de várias coordenadas. Em (distância, paradas), a ordem diz o que representa cada número. A dimensão é a quantidade de coordenadas. Para somar dois vetores ou multiplicar coordenadas correspondentes, confira dimensões iguais. A soma de (1, 2) e (3, 4) é (1+3, 2+4), ou (4, 6). No produto escalar somamos os produtos depois de multiplicar os pares: (1, 2) e (3, 4) fornecem 1×3 + 2×4 = 11. Revise [tuplas e desempacotamento](https://devrgs.github.io/Aprendendo-linguagens/#/python/registros/tuplas-e-desempacotamento).

Execute as células na ordem com **Shift+Enter**. A preparação cria uma pasta nova na área de arquivos do caderno. Baixe o caderno editado e os resultados que quiser guardar.


````python
from pathlib import Path as _Path
from tempfile import mkdtemp as _mkdtemp
import os as _os
import sys as _sys
from importlib import invalidate_caches as _refresh_imports

# Auxiliares editáveis devem ser relidos sem aproveitar bytecode antigo.
_sys.dont_write_bytecode = True

_raiz = _Path.cwd()
_trabalho = _Path(_mkdtemp(prefix='python-vetores-', dir=_raiz))
````

## E1 · Preveja e execute

Calcule os produtos antes de executar e compare o total.


````python
_pasta = _trabalho / 'e1'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
x, y = [1, 2], [3, 4]
produtos = [a * b for a, b in zip(x, y, strict=True)]
print(produtos)
print(sum(produtos))
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````


A correspondência é por posição. O produto escalar devolve um número.

## E3 · Resolva

Combine pesos 0.25, 0.75 com os valores 4, 8. Depois faça o mesmo com um elemento e com comprimentos distintos.


````python
_pasta = _trabalho / 'e3'
_pasta.mkdir(exist_ok=True)
_os.chdir(_pasta)
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
if str(_pasta) not in _sys.path:
    _sys.path.insert(0, str(_pasta))
_refresh_imports()
````

````python
def ponderar(pesos, valores):
    return sum(peso * valor for peso, valor in zip(pesos, valores, strict=True))

print(ponderar([0.25, 0.75], [4, 8]))
assert ponderar([1], [12]) == 12
try:
    ponderar([1], [2, 3])
except ValueError:
    print("tamanhos diferentes")
````

````python
# Voltar à pasta do caderno para a etapa seguinte.
_os.chdir(_raiz)
if str(_pasta) in _sys.path:
    _sys.path.remove(str(_pasta))
# Liberar módulos locais para a próxima etapa importar sua própria versão.
for _nome, _modulo in list(_sys.modules.items()):
    _origem = getattr(_modulo, '__file__', None)
    if _origem and _Path(_origem).is_relative_to(_pasta):
        del _sys.modules[_nome]
````

Saída esperada

```text
7.0
tamanhos diferentes
```

Os produtos são 1 e 6. Em uma média ponderada os pesos somam 1. O modo estrito também protege o caso de tamanhos diferentes.
